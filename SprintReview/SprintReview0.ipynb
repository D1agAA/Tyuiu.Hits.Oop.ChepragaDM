```polyglot-notebook
using System;

public interface IDryable
{
    void Dry();
}

public interface IAgeable
{
    void AgeGrow(int years);
}

public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine("Растение цветёт");
    }

    public virtual void Growing(float amount)
    {
        Height += amount;
        Console.WriteLine($"Растение выросло на {amount}. Теперь его высота: {Height}");
    }
}

public class Flower : Plant, IDryable
{
    public string Colour { get; set; }
    public int PetalNumber { get; set; }

    public Flower(string name, string type, float height, string colour, int petalnumber)
        : base(name, type, height)
    {
        Colour = colour;
        PetalNumber = petalnumber;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Цветок {Name} распустился и радует глаз");
    }

    public void Dry()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать");
    }
}

public class Tree : Plant, IAgeable
{
    public int Age { get; set; }
    public string LeavesType { get; set; }

    public Tree(string name, string type, float height, int age, string leavestype)
        : base(name, type, height)
    {
        Age = age;
        LeavesType = leavestype;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями");
    }

    public void AgeGrow(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево {Name} постарело на {years} лет, теперь его возраст {Age}");
    }
}

// ===== СОЗДАНИЕ ОБЪЕКТОВ И ВЫВОД =====

Console.WriteLine("=== Демонстрация Наследования и Интерфейсов ===");
Console.WriteLine();

Flower myFlower = new Flower("Роза", "Цветок", 0.5f, "Красный", 35);
Tree myTree = new Tree("Дуб", "Дерево", 5.0f, 10, "Зелёные");
Plant myGenericPlant = new Plant("Кактус", "Колючий", 0.2f);

myFlower.Bloom();
myFlower.Dry();

Console.WriteLine();

myTree.Bloom();
myTree.AgeGrow(5);

Console.WriteLine();

myGenericPlant.Bloom();
myGenericPlant.Growing(0.3f);

Console.WriteLine();
Console.WriteLine("=== Полиморфизм через массив ===");
Console.WriteLine();

Plant[] garden = new Plant[] { myFlower, myTree, myGenericPlant };

foreach (Plant p in garden)
{
    p.Bloom();
    p.Growing(0.1f);

    // Классический способ проверки типа (работает везде)
    if (p is IDryable)
    {
        IDryable dryable = (IDryable)p;
        dryable.Dry();
    }

    if (p is IAgeable)
    {
        IAgeable ageable = (IAgeable)p;
        ageable.AgeGrow(1);
    }

    Console.WriteLine();
}
```
