```polyglot-notebook
public abstract class Component
{
    public string Model {get;set;}
    public double Price {get;set;}
    public Component(string model, double price)
    {
        Model = model;
        Price = price;
    }
    public virtual string GetInfo()
    {
        return $"Компонент: {Model}, Цена: {Price}";
    }
    public override  string ToString()
    {
        return GetInfo();
    }
    public static bool operator ==(Component left, Component right)
    {
        if(ReferenceEquals(left, null) && ReferenceEquals(right, null))
            return true;
        if(ReferenceEquals(left,null) || ReferenceEquals(right,null))
            return false;
        return left.Model == right.Model &&  left.Price == right.Price;
    }
    public static bool operator !=(Component left, Component right)
    {
        return !(left == right);
    }
    public override bool Equals(object obj)
    {
        if (obj is Component other)
            return this == other;
        return false;
    }
    public override int GetHashCode()
    {
        return HashCode.Combine(Model, Price);
    }
}
public interface IInstallsble
{
    void Install();
}
public class CPU : Component, IInstallsble
{
    public double Cores {get;set;}
    public CPU(string model, double price, double cores) : base(model,price)
    {
        Cores = cores;
    }
    public override string GetInfo()
    {
        return $"[CPU] Модель {Model}, Цена {Price}, {Cores} ядер";
    }
    public void Install()
    {
        Console.WriteLine("Идёт установка CPU в сокет мат. платы");
        Console.WriteLine("Установка выполнена");
    }
}
public class RAM : Component, IInstallsble
{
    public int Capacity {get;set;}

    public RAM(string model, double price, int capacity) : base(model,price)
    {
        Capacity = capacity;
    }
    public override string GetInfo()
    {
        return $"[RAM], Модель {Model}, Цена {Price}, Кол-во памяти {Capacity} ГБ";
    }
    public void Install()
    {
        Console.WriteLine("Идёт установка плашки RAM в слот DDIM");
        Console.WriteLine("Установка выполнена");
    }
}
public class GPU : Component, IInstallsble
{
    public int Memory {get;set;}

    public GPU(string model, double price, int memory) : base(model,price)
    {
        Memory = memory;
    }
    public override string GetInfo()
    {
        return $"[GPU], Модель {Model}, Цена {Price}, Видеопамять {Memory}ГБ";
    }
    public void Install()
    {
        Console.WriteLine("Идёт установка GPU в слот PCI-e");
        Console.WriteLine("Установка прошла успешно");
    }
}
public class HDD : Component, IInstallsble
{
    public int Capacity {get;set;}

    public HDD(string model, double price, int capacity) : base(model,price)
    {
        Capacity = capacity;
    }
    public override string GetInfo()
    {
        return $"[HDD], Модель {Model}, Цена {Price}, Вместимость {Capacity} ГБ";
    }
    public void Install()
    {
        Console.WriteLine("Идёт установка HDD в отсек и подключение через SATA");
        Console.WriteLine("Установка прошла успешно");
    }
}
public class Computer
{
    private List<Component> _components;
    private string _name;

    public Computer(string name)
    {
        _name = name;
        _components = new List<Component>();
    }
    public void AddComponent<T>(T component) where  T : Component, IInstallsble
    {
        _components.Add(component);
        Console.WriteLine($"[{_name}] Добавлен компонент : {component.Model}");
        component.Install();
    }
    public void RemoveComponent<T> (T component) where T : Component
    {
        if (_components.Remove(component))
        {
            Console.WriteLine($"[{_name}] Удалён компонент : {component.Model}");
        }
        else
        {
            Console.WriteLine($"[{_name}] компонент {component.Model} не найден");
        }
    }
    public double GetTotalPrice()
    {
        double total = 0;
        foreach (var v in _components)
        {
            total += v.Price;
        }
        return total;
    }
    public void ShowComponents()
    {
        Console.WriteLine($"\n[{_name}] Конфигурация компьютера:");
        foreach (var component in _components)
        {
            Console.WriteLine($"  - {component.GetInfo()}");
        }
        Console.WriteLine($"  Итого: {GetTotalPrice()} руб.");
    }
}
public static class ComputerFactory
{
    public static Computer CreateGamingPC()
    {
        Computer pc = new Computer("Игровой ПК");
        pc.AddComponent(new CPU("Intel Core i9-13900K", 45000, 24));
        pc.AddComponent(new RAM("Kingston Fury 32GB", 15000, 32));
        pc.AddComponent(new GPU("NVIDIA RTX 4080", 95000, 16));
        pc.AddComponent(new HDD("Samsung 980 Pro 2TB", 18000, 2000));
        return pc;
    }
    public static Computer CreateOfficePC()
    {
        Computer pc = new Computer("Офисный ПК");
        pc.AddComponent(new CPU("Intel Core i5-12400", 18000, 6));
        pc.AddComponent(new RAM("Kingston Fury 16GB", 6000, 16));
        pc.AddComponent(new GPU("Intel UHD 730", 0, 0));
        pc.AddComponent(new HDD("WD Blue 1TB", 4500, 1000));
        return pc;
    }
}
Computer gamingPC = ComputerFactory.CreateGamingPC();
gamingPC.ShowComponents();

Computer officePC = ComputerFactory.CreateOfficePC();
officePC.ShowComponents();

Console.WriteLine("\n=== Демонстрация удаления компонента ===");
gamingPC.RemoveComponent(new GPU("NVIDIA RTX 4080", 95000, 16));
gamingPC.ShowComponents();

Console.WriteLine("\n=== Демонстрация сравнения компонентов ===");
Component cpu1 = new CPU("Intel Core i9", 45000, 24);
Component cpu2 = new CPU("Intel Core i9", 45000, 24);
Component cpu3 = new CPU("AMD Ryzen 9", 55000, 16);

Console.WriteLine($"cpu1 == cpu2: {cpu1 == cpu2}");
Console.WriteLine($"cpu1 != cpu3: {cpu1 != cpu3}");

Console.WriteLine("\n=== Демонстрация полиморфизма ===");
List<Component> allComponents = new List<Component>
{
    new CPU("Test CPU", 10000, 4),
    new RAM("Test RAM", 5000, 8),
    new GPU("Test GPU", 20000, 8),
    new HDD("Test HDD", 3000, 500)
};

foreach (var component in allComponents)
{
    Console.WriteLine(component.GetInfo());
}
```
