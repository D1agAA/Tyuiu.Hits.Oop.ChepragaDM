----

Программа на C# демонстрирует базовые принципы ООП:
1. Инкапсуляция — скрытие данных с помощью свойств.
2. Наследование — создание классов Email, SMS, Push на основе базового класса Notification.
3. Полиморфизм — переопределение метода SendNotification в каждом классе.
4. Абстракция — использование абстрактного класса для задания общего шаблона.

Также показано разделение обязанностей (принцип SRP): 
базовый класс отвечает за общие данные, а наследники — за специфику отправки.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

public abstract class Notification
{
    private int _id;
    private string _message;

    public int Id 
    { 
        get { return _id; } 
        set { _id = value; } 
    }

    public string Message 
    { 
        get { return _message; } 
        set { _message = value; } 
    }

    public Notification(int id, string message)
    {
        Id = id;
        Message = message;
    }

    public abstract void SendNotification();

    public void DisplayNotification()
    {
        Console.WriteLine($"Уведомление #{Id}: {Message}");
    }
}

public class EmailNotification : Notification
{
    public string EmailAddress { get; set; }

    public EmailNotification(int id, string message, string email) 
        : base(id, message)
    {
        EmailAddress = email;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"[Email] Отправка на {EmailAddress}: {Message}");
    }
}

public class SMSNotification : Notification
{
    public string PhoneNumber { get; set; }

    public SMSNotification(int id, string message, string phone) 
        : base(id, message)
    {
        PhoneNumber = phone;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"[SMS] Отправка на номер {PhoneNumber}: {Message}");
    }
}

public class PushNotification : Notification
{
    public string Platform { get; set; }

    public PushNotification(int id, string message, string platform) 
        : base(id, message)
    {
        Platform = platform;
    }

    public override void SendNotification()
    {
        Console.WriteLine($"[Push] Отправка на платформу {Platform}: {Message}");
    }
}

EmailNotification email = new EmailNotification(1, "Ваш заказ отправлен", "client@mail.ru");
SMSNotification sms = new SMSNotification(2, "Код подтверждения: 5678", "+7-999-123-45-67");
PushNotification push = new PushNotification(3, "Новое сообщение!", "Android");

Console.WriteLine("--- Email ---");
email.DisplayNotification();
email.SendNotification();

Console.WriteLine();
Console.WriteLine("--- SMS ---");
sms.DisplayNotification();
sms.SendNotification();

Console.WriteLine();
Console.WriteLine("--- Push ---");
push.DisplayNotification();
push.SendNotification();
```
