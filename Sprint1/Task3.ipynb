Программа на C# демонстрирует базовые принципы ООП:
1. Инкапсуляция — скрытие данных с помощью свойств.
2. Наследование — создание классов Email, SMS, Push на основе базового класса Notification.
3. Полиморфизм — переопределение метода SendNotification в каждом классе.
4. Абстракция — использование абстрактного класса для задания общего шаблона.

Также показано разделение обязанностей (принцип SRP): 
базовый класс отвечает за общие данные, а наследники — за специфику отправки.

```
using System;
using System.Collections.Generic;

public abstract class Notification
{

    private int _id;
    private string _message;
    private string _type;

    public int Id
    {
        get { return _id; }
        set
        {
            if (value > 0)
                _id = value;
            else
                throw new ArgumentOutOfRangeException("ID должен быть положительным!");
        }
    }

    public string Message
    {
        get { return _message; }
        set
        {
            if (!string.IsNullOrEmpty(value))
                _message = value;
            else
                throw new ArgumentOutOfRangeException("Сообщение не может быть пустым!");
        }
    }

    public string Type
    {
        get { return _type; }
        set { _type = value; }
    }

    public static int TotalNotifications { get; private set; } = 0;

    public Notification(int id, string message, string type)
    {
        Id = id;
        Message = message;
        Type = type;
        TotalNotifications++; 
    }

    public abstract void SendNotification();

    // Обычный метод
    public void DisplayNotification()
    {
        Console.WriteLine($"Уведомление #{Id} [{Type}]: {Message}");
    }

    public static void DisplayTotalNotifications()
    {
        Console.WriteLine($"Всего уведомлений создано: {TotalNotifications}");
    }
}

public class EmailNotification : Notification
{

    private string _emailAddress;
    private string _subject;


    public string EmailAddress
    {
        get { return _emailAddress; }
        set
        {
            if (value.Contains("@"))
                _emailAddress = value;
            else
                throw new ArgumentException("Некорректный email!");
        }
    }

    public string Subject
    {
        get { return _subject; }
        set { _subject = value; }
    }

   
    public EmailNotification(int id, string message, string email, string subject)
        : base(id, message, "Email")
    {
        EmailAddress = email;
        Subject = subject;
    }

   
    public override void SendNotification()
    {
        Console.WriteLine($"[Email] Кому: {EmailAddress}");
        Console.WriteLine($"        Тема: {Subject}");
        Console.WriteLine($"        Текст: {Message}");
        Console.WriteLine($"        Статус: ✓ Отправлено");
    }
}

public class SMSNotification : Notification
{
    private string _phoneNumber;
    private int _characterCount;

    public string PhoneNumber
    {
        get { return _phoneNumber; }
        set
        {
            if (value.StartsWith("+") || value.StartsWith("8"))
                _phoneNumber = value;
            else
                throw new ArgumentException("Некорректный номер телефона!");
        }
    }

    public int CharacterCount
    {
        get { return _characterCount; }
        set { _characterCount = value; }
    }


    public SMSNotification(int id, string message, string phone)
        : base(id, message, "SMS")
    {
        PhoneNumber = phone;
        CharacterCount = message.Length;  
    }

    public override void SendNotification()
    {
        Console.WriteLine($"[SMS] Кому: {PhoneNumber}");
        Console.WriteLine($"      Текст ({CharacterCount} симв.): {Message}");
        Console.WriteLine($"      Статус: ✓ Доставлено");
    }
}

public class PushNotification : Notification
{
    private string _platform;
    private string _icon;

    public string Platform
    {
        get { return _platform; }
        set
        {
            if (value == "iOS" || value == "Android")
                _platform = value;
            else
                throw new ArgumentException("Платформа должна быть iOS или Android!");
        }
    }

    public string Icon
    {
        get { return _icon; }
        set { _icon = value; }
    }

    public PushNotification(int id, string message, string platform)
        : base(id, message, "Push")
    {
        Platform = platform;
        Icon = platform == "iOS" ? "🍎" : "🤖";
    }

    public override void SendNotification()
    {
        Console.WriteLine($"{Icon} [Push] Платформа: {Platform}");
        Console.WriteLine($"        Текст: {Message}");
        Console.WriteLine($"        Статус: ✓ Доставлено");
    }
}

public class NotificationManager
{
    private string _managerName;
    private List<Notification> _notifications;  

    public string ManagerName
    {
        get { return _managerName; }
        set { _managerName = value; }
    }


    public NotificationManager(string managerName)
    {
        ManagerName = managerName;
        _notifications = new List<Notification>();
    }

    public void AddNotification(Notification notification)
    {
        _notifications.Add(notification);
        Console.WriteLine($"[Менеджер {ManagerName}] принял уведомление #{notification.Id}");
    }

    public void SendAll()
    {
        Console.WriteLine($"\n[Менеджер {ManagerName}] начинает рассылку...\n");
        foreach (Notification n in _notifications)
        {
            n.SendNotification();
            Console.WriteLine();
        }
    }

    public void ShowAll()
    {
        Console.WriteLine($"[Менеджер {ManagerName}] список уведомлений:");
        foreach (Notification n in _notifications)
        {
            n.DisplayNotification();
        }
    }
}


Console.WriteLine("=== СИСТЕМА УВЕДОМЛЕНИЙ ===\n");

EmailNotification email = new EmailNotification(
    1,
    "Ваш заказ #1234 отправлен",
    "client@mail.ru",
    "Заказ отправлен"
);

SMSNotification sms = new SMSNotification(
    2,
    "Код подтверждения: 5678",
    "+7-999-123-45-67"
);

PushNotification push = new PushNotification(
    3,
    "У вас новое сообщение!",
    "Android"
);

Console.WriteLine("--- Прямая работа с уведомлениями ---\n");

Notification[] allNotifications = { email, sms, push };

foreach (Notification n in allNotifications)
{
    n.DisplayNotification();
    n.SendNotification();
    Console.WriteLine();
}

Console.WriteLine("\n--- Взаимодействие через менеджера ---\n");

NotificationManager manager = new NotificationManager("Иван");

manager.AddNotification(email);
manager.AddNotification(sms);
manager.AddNotification(push);

manager.ShowAll();

manager.SendAll();

Notification.DisplayTotalNotifications();
```
