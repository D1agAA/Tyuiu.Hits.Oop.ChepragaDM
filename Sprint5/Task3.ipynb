Программа на C# демонстрирует базовые принципы ООП:
1. Инкапсуляция — скрытие данных с помощью свойств.
2. Наследование — создание классов Email, SMS, Push на основе базового класса Notification.
3. Полиморфизм — переопределение метода SendNotification в каждом классе.
4. Абстракция — использование абстрактного класса для задания общего шаблона.

Также показано разделение обязанностей (принцип SRP): 
базовый класс отвечает за общие данные, а наследники — за специфику отправки.

```
using System;
using System.Collections.Generic;
using System.Linq;

delegate void AnimalProcessor(Animal a);

interface IAnimal { void Move(); void Eat(); int GetLegs(); }
interface IPet { string Owner { get; set; } void Play(); }

class Animal : IAnimal
{
    private string _name;
    public string Name
    {
        get => _name;
        set => _name = !string.IsNullOrEmpty(value) ? value : throw new ArgumentException("Имя пустое!");
    }

    private int _age;
    public int Age
    {
        get => _age;
        set => _age = value >= 0 ? value : throw new ArgumentOutOfRangeException("Возраст < 0!");
    }

    public string Habitat { get; set; } = "неизвестно";
    public double Weight { get; set; }
    public bool IsHungry { get; set; } = true;
    public string Color { get; set; } = "неизвестен";

    public delegate void AnimalEventHandler(string message);
    public event AnimalEventHandler AnimalAdded;
    public event AnimalEventHandler NameChanged;

    public Animal(string name, int age) { Name = name; Age = age; }

    void IAnimal.Move() => Console.WriteLine($"[явно] {Name} двигается");
    void IAnimal.Eat() { Console.WriteLine($"[явно] {Name} кушает"); IsHungry = false; }
    int IAnimal.GetLegs() => 4;

    public virtual void Move() => Console.WriteLine($"{Name} двигается");
    public virtual void Eat() { Console.WriteLine($"{Name} кушает"); IsHungry = false; }
    public virtual int GetLegs() => 4;
    public virtual void Sleep() => Console.WriteLine($"{Name} спит");

    public void Eat(string food) => Console.WriteLine($"{Name} кушает {food}");
    public void Eat(string food, int portions) => Console.WriteLine($"{Name} кушает {food} x{portions}");

    public virtual string Describe(bool shortForm) => shortForm
        ? $"{Name} ({Age} г.)"
        : $"{Name}, {Age} г., окрас: {Color}, вес: {Weight} кг, среда: {Habitat}";

    public virtual string GetInfo() => $"Животное: {Name}, Возраст: {Age}, Среда: {Habitat}";

    public void OnAdded() => AnimalAdded?.Invoke($"Добавлено: {Name}");
    public void Rename(string newName) { Name = newName; NameChanged?.Invoke($"Имя изменено на {newName}"); }
}

class Dog : Animal, IPet
{
    public string Breed { get; set; }
    public bool IsTrained { get; set; }
    public string FavoriteToy { get; set; } = "мяч";
    public string Owner { get; set; } = "не указан";

    public Dog(string name, int age, string breed) : base(name, age) { Breed = breed; Habitat = "двор"; }

    public void Play() => Console.WriteLine($"{Name} играет с {FavoriteToy}");
    public override void Move() => Console.WriteLine($"{Name} ({Breed}) бегает");
    public override void Eat() { Console.WriteLine($"{Name} грызет косточку"); IsHungry = false; }
    public override void Sleep() => Console.WriteLine($"{Name} спит в будке");
    public void Fetch() => Console.WriteLine($"{Name} приносит {FavoriteToy}");
    public void Eat(string food, bool fromOwner) => Console.WriteLine($"{Name} кушает {food} {(fromOwner ? "из рук" : "из миски")}");

    public override string Describe(bool shortForm) => shortForm
        ? $"Собака {Name} ({Breed})"
        : base.Describe(shortForm) + $", дрессирована: {(IsTrained ? "Да" : "Нет")}, игрушка: {FavoriteToy}, хозяин: {Owner}";
}

class Cat : Animal, IPet
{
    public bool IsFluffy { get; set; }
    public bool IsIndoor { get; set; } = true;
    public int LivesLeft { get; set; } = 9;
    public string Owner { get; set; } = "не указан";

    public Cat(string name, int age, bool isFluffy) : base(name, age) { IsFluffy = isFluffy; Habitat = "дом"; }

    public void Play() => Console.WriteLine($"{Name} играет с клубком");
    public override void Move() => Console.WriteLine($"{Name} крадется");
    public override void Eat() { Console.WriteLine($"{Name} лакает молоко"); IsHungry = false; }
    public override void Sleep() => Console.WriteLine($"{Name} спит на подоконнике");
    public void Purr() => Console.WriteLine($"{Name} мурлычет");
    public void Move(bool silently) => Console.WriteLine($"{Name} {(silently ? "крадется бесшумно" : "топает")}");

    public override string Describe(bool shortForm) => shortForm
        ? $"Кошка {Name}"
        : base.Describe(shortForm) + $", домашняя: {(IsIndoor ? "Да" : "Нет")}, жизней: {LivesLeft}, хозяин: {Owner}";
}

class Bird : Animal
{
    public double WingSpan { get; set; }
    public bool CanFly { get; set; } = true;
    public string Species { get; set; } = "неизвестен";

    public Bird(string name, int age, double wingSpan) : base(name, age) { WingSpan = wingSpan; Habitat = "лес"; }

    public override void Move() => Console.WriteLine($"{Name} летает ({WingSpan} м)");
    public override void Eat() { Console.WriteLine($"{Name} клюет зерна"); IsHungry = false; }
    public override int GetLegs() => 2;
    public override void Sleep() => Console.WriteLine($"{Name} спит в гнезде");
    public void Sing() => Console.WriteLine($"{Name} поёт");
    public void Move(double altitude) => Console.WriteLine(CanFly
        ? $"{Name} летает на {altitude} м"
        : $"{Name} не умеет летать");

    public override string Describe(bool shortForm) => shortForm
        ? $"Птица {Name}"
        : base.Describe(shortForm) + $", вид: {Species}, летает: {(CanFly ? "Да" : "Нет")}";
}

class AnimalCollection<T> where T : Animal
{
    private readonly List<T> _animals = new();
    public int Count => _animals.Count;
    public T this[int i] => _animals[i];

    public void Add(T a) { if (a == null) throw new ArgumentNullException(nameof(a)); _animals.Add(a); }
    public bool Remove(T a) => _animals.Remove(a);
    public void MoveAll() { foreach (var a in _animals) a.Move(); }
    public void SleepAll() { foreach (var a in _animals) a.Sleep(); }
    public void DisplayAll() { foreach (var a in _animals) Console.WriteLine(a.Describe(false)); }
    public void FeedAll() { foreach (var a in _animals) a.Eat(); }
    public void FeedAll(string food) { foreach (var a in _animals) a.Eat(food); }
    public void FeedAll(string food, int portions) { foreach (var a in _animals) a.Eat(food, portions); }
}

class AnimalService
{
    private readonly IAnimal _a;
    public AnimalService(IAnimal a) => _a = a;
    public void UseAnimal() { _a.Move(); Console.WriteLine($"Ног: {_a.GetLegs()}"); _a.Eat(); }
}

class PetService
{
    private readonly IPet _p;
    public PetService(IPet p) => _p = p;
    public void UsePet() { Console.WriteLine($"Хозяин: {_p.Owner}"); _p.Play(); }
}

var dog = new Dog("Шарик", 3, "Лабрадор") { IsTrained = true, FavoriteToy = "кость", Owner = "Иван" };
var cat = new Cat("Мурка", 2, true) { Owner = "Мария" };
var bird = new Bird("Кеша", 1, 0.5) { Species = "попугай" };

var animalList = new List<Animal>();
var animalDict = new Dictionary<string, Animal>();

Action<string, Animal> addAnimal = (id, a) =>
{
    animalList.Add(a); animalDict[id] = a;
    a.AnimalAdded += Console.WriteLine;
    a.NameChanged += Console.WriteLine;
    a.OnAdded();
};

addAnimal("D01", dog);
addAnimal("C01", cat);
addAnimal("B01", bird);
addAnimal("B02", new Bird("Пит", 2, 0.4) { CanFly = false, Species = "пингвин" });

Console.WriteLine("=== Делегат: обход ===");
AnimalProcessor processor = a => a.Move();
foreach (var a in animalList) processor(a);

Console.WriteLine("\n=== Лямбда + LINQ: младше 3 лет ===");
animalList.Where(a => a.Age < 3).ToList().ForEach(a => Console.WriteLine(a.Describe(true)));

Console.WriteLine("\n=== Dictionary: поиск по ID ===");
if (animalDict.TryGetValue("C01", out Animal found)) Console.WriteLine(found.Describe(false));

Console.WriteLine("\n=== Событие: переименование ===");
animalDict["D01"].Rename("Барбос");

Console.WriteLine("\n=== Сортировка (лямбда-компаратор) ===");
animalList.Sort((x, y) => x.Age.CompareTo(y.Age));
animalList.ForEach(a => Console.WriteLine(a.Describe(true)));

Console.WriteLine("\n=== Перегрузки Eat ===");
dog.Eat(); dog.Eat("косточку"); dog.Eat("косточку", 2); dog.Eat("косточку", true);

Console.WriteLine("\n=== Перегрузки Move ===");
cat.Move(); cat.Move(true); bird.Move(); bird.Move(100.0);

Console.WriteLine("\n=== IPet через PetService ===");
new PetService(dog).UsePet();
new PetService(cat).UsePet();

Console.WriteLine("\n=== IAnimal через AnimalService ===");
new AnimalService(dog).UseAnimal();
new AnimalService(bird).UseAnimal();

Console.WriteLine($"\nВсего: {animalList.Count}, первый: {animalList[0].Describe(true)}");
```
