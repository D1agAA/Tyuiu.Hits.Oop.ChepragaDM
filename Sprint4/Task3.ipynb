```
using System;
using System.Collections.Generic;

public interface IAnimal
{
    void Move();
    void Eat();
    int GetLegs();
}

public interface IPet
{
    string Owner { get; set; }
    void Play();
}

public class Animal : IAnimal
{
    private string _name;
    public string Name
    {
        get { return _name; }
        set
        {
            if (!string.IsNullOrEmpty(value)) _name = value;
            else throw new ArgumentException("Имя не может быть пустым!");
        }
    }

    private int _age;
    public int Age
    {
        get { return _age; }
        set
        {
            if (value >= 0) _age = value;
            else throw new ArgumentOutOfRangeException("Возраст не может быть отрицательным!");
        }
    }

    public string Habitat { get; set; }
    public double Weight { get; set; }
    public bool IsHungry { get; set; }
    public string Color { get; set; }

    public Animal(string name, int age)
    {
        Name = name;
        Age = age;
        Habitat = "неизвестно";
        Weight = 0;
        IsHungry = true;
        Color = "неизвестен";
    }

    void IAnimal.Move()
    {
        Console.WriteLine($"[явно] {Name} двигается");
    }

    void IAnimal.Eat()
    {
        Console.WriteLine($"[явно] {Name} кушает");
        IsHungry = false;
    }

    int IAnimal.GetLegs() => 4;

    public virtual void Move()
    {
        Console.WriteLine($"{Name} двигается");
    }

    public virtual void Eat()
    {
        Console.WriteLine($"{Name} кушает");
        IsHungry = false;
    }

    public virtual int GetLegs() => 4;

    public virtual string GetInfo()
    {
        return $"Животное: {Name}, Возраст: {Age}, Среда: {Habitat}";
    }

    public void Eat(string food)
    {
        Console.WriteLine($"{Name} кушает {food}");
    }

    public void Eat(string food, int portions)
    {
        Console.WriteLine($"{Name} кушает {food} в количестве {portions} порций");
    }

    public virtual void Sleep()
    {
        Console.WriteLine($"{Name} спит");
    }

    public virtual string Describe(bool shortForm)
    {
        return shortForm
            ? $"{Name} ({Age} г.)"
            : $"{Name}, {Age} г., окрас: {Color}, вес: {Weight} кг, среда: {Habitat}";
    }
}

public class Dog : Animal, IPet
{
    public string Breed { get; set; }
    public bool IsTrained { get; set; }
    public string FavoriteToy { get; set; }
    public string Owner { get; set; }

    public Dog(string name, int age, string breed) : base(name, age)
    {
        Breed = breed;
        IsTrained = false;
        FavoriteToy = "мяч";
        Owner = "не указан";
        Habitat = "двор";
    }

    public void Play()
    {
        Console.WriteLine($"{Name} играет с {FavoriteToy}");
    }

    public override void Move()
    {
        Console.WriteLine($"{Name} (порода {Breed}) бегает на четырех лапах");
    }

    public override void Eat()
    {
        Console.WriteLine($"{Name} грызет косточку");
        IsHungry = false;
    }

    public override int GetLegs() => 4;

    public override string GetInfo()
    {
        return base.GetInfo() + $", Порода: {Breed}";
    }

    public void Fetch()
    {
        Console.WriteLine($"{Name} приносит {FavoriteToy}");
    }

    public override void Sleep()
    {
        Console.WriteLine($"{Name} спит в будке, свернувшись калачиком");
    }

    public void Eat(string food, bool fromOwner)
    {
        Console.WriteLine($"{Name} кушает {food} {(fromOwner ? "из рук хозяина" : "из миски")}");
    }

    public override string Describe(bool shortForm)
    {
        return shortForm
            ? $"Собака {Name} ({Breed})"
            : base.Describe(shortForm) + $", дрессирована: {(IsTrained ? "Да" : "Нет")}, игрушка: {FavoriteToy}, хозяин: {Owner}";
    }
}

public class Cat : Animal, IPet
{
    public bool IsFluffy { get; set; }
    public bool IsIndoor { get; set; }
    public int LivesLeft { get; set; }
    public string Owner { get; set; }

    public Cat(string name, int age, bool isFluffy) : base(name, age)
    {
        IsFluffy = isFluffy;
        IsIndoor = true;
        LivesLeft = 9;
        Owner = "не указан";
        Habitat = "дом";
    }

    public void Play()
    {
        Console.WriteLine($"{Name} играет с клубком");
    }

    public override void Move()
    {
        Console.WriteLine($"{Name} мягко крадется");
    }

    public override void Eat()
    {
        Console.WriteLine($"{Name} лакает молоко");
        IsHungry = false;
    }

    public override int GetLegs() => 4;

    public override string GetInfo()
    {
        return base.GetInfo() + $", Пушистый: {(IsFluffy ? "Да" : "Нет")}";
    }

    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет");
    }

    public override void Sleep()
    {
        Console.WriteLine($"{Name} спит на подоконнике");
    }

    public void Move(bool silently)
    {
        Console.WriteLine(silently
            ? $"{Name} крадется совершенно бесшумно"
            : $"{Name} громко топает");
    }

    public override string Describe(bool shortForm)
    {
        return shortForm
            ? $"Кошка {Name}"
            : base.Describe(shortForm) + $", домашняя: {(IsIndoor ? "Да" : "Нет")}, жизней: {LivesLeft}, хозяин: {Owner}";
    }
}

public class Bird : Animal
{
    public double WingSpan { get; set; }
    public bool CanFly { get; set; }
    public string Species { get; set; }

    public Bird(string name, int age, double wingSpan) : base(name, age)
    {
        WingSpan = wingSpan;
        CanFly = true;
        Species = "неизвестен";
        Habitat = "лес";
    }

    public override void Move()
    {
        Console.WriteLine($"{Name} летает (размах крыльев: {WingSpan} м)");
    }

    public override void Eat()
    {
        Console.WriteLine($"{Name} клюет зернышки");
        IsHungry = false;
    }

    public override int GetLegs() => 2;

    public override string GetInfo()
    {
        return base.GetInfo() + $", Размах крыльев: {WingSpan} м";
    }

    public void Sing()
    {
        Console.WriteLine($"{Name} поёт");
    }

    public override void Sleep()
    {
        Console.WriteLine($"{Name} спит в гнезде");
    }

    public void Move(double altitude)
    {
        Console.WriteLine(CanFly
            ? $"{Name} летает на высоте {altitude} м"
            : $"{Name} не умеет летать (высота {altitude} м недоступна)");
    }

    public override string Describe(bool shortForm)
    {
        return shortForm
            ? $"Птица {Name}"
            : base.Describe(shortForm) + $", вид: {Species}, летает: {(CanFly ? "Да" : "Нет")}";
    }
}

public class AnimalCollection<T> where T : Animal
{
    private readonly List<T> _animals = new List<T>();

    public int Count => _animals.Count;
    public T this[int index] => _animals[index];

    public void Add(T animal)
    {
        if (animal == null) throw new ArgumentNullException(nameof(animal));
        _animals.Add(animal);
    }

    public bool Remove(T animal) => _animals.Remove(animal);

    public void MoveAll()
    {
        foreach (var a in _animals) a.Move();
    }

    public void SleepAll()
    {
        foreach (var a in _animals) a.Sleep();
    }

    public void DisplayAll()
    {
        foreach (var a in _animals)
            Console.WriteLine(a.Describe(false));
    }

    public void FeedAll()
    {
        foreach (var a in _animals) a.Eat();
    }

    public void FeedAll(string food)
    {
        foreach (var a in _animals) a.Eat(food);
    }

    public void FeedAll(string food, int portions)
    {
        foreach (var a in _animals) a.Eat(food, portions);
    }
}

public class AnimalService
{
    private readonly IAnimal _animal;

    public AnimalService(IAnimal animal)
    {
        _animal = animal;
    }

    public void UseAnimal()
    {
        _animal.Move();
        Console.WriteLine($"Ног: {_animal.GetLegs()}");
        _animal.Eat();
    }
}

public class PetService
{
    private readonly IPet _pet;

    public PetService(IPet pet)
    {
        _pet = pet;
    }

    public void UsePet()
    {
        Console.WriteLine($"Хозяин: {_pet.Owner}");
        _pet.Play();
    }
}

Console.WriteLine("========== ДОП. ЗАДАНИЕ (на основе Task0–Task3) ==========");
Console.WriteLine();

Dog dog = new Dog("Шарик", 3, "Лабрадор");
dog.IsTrained = true;
dog.FavoriteToy = "резиновая кость";
dog.Owner = "Иван";

Cat cat = new Cat("Мурка", 2, true);
cat.IsIndoor = true;
cat.Owner = "Мария";

Bird bird = new Bird("Кеша", 1, 0.5);
bird.Species = "попугай";

Console.WriteLine("--- 1. Перекрытие Move / Sleep ---");
Animal[] animals = { dog, cat, bird };
foreach (var a in animals)
{
    a.Move();
    a.Sleep();
    Console.WriteLine();
}

Console.WriteLine("--- 2. Перегрузка Eat ---");
dog.Eat();
dog.Eat("косточку");
dog.Eat("косточку", 2);
dog.Eat("косточку", true);
Console.WriteLine();

Console.WriteLine("--- 3. Перегрузка Move в наследниках ---");
cat.Move();
cat.Move(true);
bird.Move();
bird.Move(100.0);
Console.WriteLine();

Console.WriteLine("--- 4. Перекрытие Describe ---");
foreach (var a in animals)
{
    Console.WriteLine("short: " + a.Describe(true));
    Console.WriteLine("full:  " + a.Describe(false));
    Console.WriteLine();
}

Console.WriteLine("--- 5. Специфичные новые методы ---");
dog.Fetch();
cat.Purr();
bird.Sing();
Console.WriteLine();

Console.WriteLine("--- 6. Generic: AnimalCollection<Dog> ---");
AnimalCollection<Dog> dogs = new AnimalCollection<Dog>();
dogs.Add(new Dog("Рекс", 5, "Овчарка"));
dogs.Add(new Dog("Бим", 4, "Дворняга"));
dogs.DisplayAll();
Console.WriteLine();
dogs.MoveAll();
dogs.FeedAll("мясо", 2);
Console.WriteLine();

Console.WriteLine("--- 7. Generic: AnimalCollection<Animal> ---");
AnimalCollection<Animal> zoo = new AnimalCollection<Animal>();
zoo.Add(dog);
zoo.Add(cat);
zoo.Add(bird);
zoo.Add(new Bird("Пингвин Пит", 2, 0.4) { CanFly = false, Species = "пингвин" });

zoo.DisplayAll();
Console.WriteLine();
zoo.MoveAll();
Console.WriteLine();
zoo.SleepAll();
Console.WriteLine();

Console.WriteLine("--- 8. Перегрузка FeedAll в generic-классе ---");
zoo.FeedAll();
zoo.FeedAll("зерно");
zoo.FeedAll("зерно", 3);
Console.WriteLine();

Console.WriteLine("--- 9. Явная реализация IAnimal ---");
IAnimal[] asAnimals = { dog, cat, bird };
foreach (var a in asAnimals)
{
    a.Move();
    a.Eat();
    Console.WriteLine($"Ног: {a.GetLegs()}");
    Console.WriteLine();
}

Console.WriteLine("--- 10. Управление зависимостями: AnimalService ---");
new AnimalService(dog).UseAnimal();
Console.WriteLine();
new AnimalService(cat).UseAnimal();
Console.WriteLine();
new AnimalService(bird).UseAnimal();
Console.WriteLine();

Console.WriteLine("--- 11. Управление зависимостями: PetService ---");
new PetService(dog).UsePet();
Console.WriteLine();
new PetService(cat).UsePet();
Console.WriteLine();

Console.WriteLine($"Всего в зоопарке: {zoo.Count}");
Console.WriteLine($"Первый элемент: {zoo[0].Describe(true)}");
```
